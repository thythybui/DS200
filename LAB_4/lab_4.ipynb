##### Câu 4

```python
orders_time = orders \
    .withColumn(
        "Year",
        year("Order_Purchase_Timestamp")
    ) \
    .withColumn(
        "Month",
        month("Order_Purchase_Timestamp")
    )

orders_time.show(5)
```

```python
orders_year_month = orders_time \
    .groupBy(
        "Year",
        "Month"
    ) \
    .agg(
        count("Order_ID")
        .alias("Total_Order")
    )\
    .orderBy(
        col("Year").asc(),
        col("Month").desc()
    )\
    .show()
```

##### Câu 5

```python
clean_reviews = order_reviews.filter(
    col("Review_Score").isNotNull()
    &
    (col("Review_Score") >= 1)
    &
    (col("Review_Score") <= 5)
)
```

```python
average_score = clean_reviews \
    .select(
        avg("Review_Score")
        .alias("Average_Score")
    )

average_score.show()
```

```python
clean_reviews.groupby("Review_Score").agg(count("Review_ID").alias("Total_Reviews")).orderBy(col("Review_Score").desc()).show()
```

##### Câu 6

```python
sales = orders \
    .join(
        order_items,
        "Order_ID"
    ) \
    .join(
        products,
        "Product_ID"
    )
```

```python
sales_2024 = sales.filter(year("Order_Purchase_Timestamp") == 2024)
```

```python
sales_2024.groupBy("Product_Category_Name").agg(sum(col("Price")+col("Freight_Value")).alias("Total_Sales")).orderBy(col("Total_Sales").desc()).show()    
```

##### Câu 7

```python
products_sale_max = products.join(
    order_items,
    "Product_ID"
).groupBy("Product_ID"
).agg(count("Order_ID"
            ).alias("Total_Sales")
            ).orderBy(col("Total_Sales"
                          ).desc())

products_sale_max.show()  
```

```python
products_avg_reviews = order_items.join(
    order_reviews,
    "Order_ID"
).groupBy("Product_ID").agg(avg("Review_Score").alias("Average_Review_Score"))

products_avg_reviews.show()
```

```python
products_sale_max_avg_review = products_sale_max.join(
    products_avg_reviews,
    "Product_ID"
)
products_sale_max_avg_review.orderBy(col("Total_Sales").desc()).show()
```

##### Câu 8

```python
delivery_time = orders.join(
    order_items,
    "Order_ID"
)\
.withColumn(
    "Delivery_Delay",
    datediff(
        col("Order_Delivered_Carrier_Date"),
        col("Shipping_Limit_Date")
    )
)\
.withColumn(
    "Status",
    when(
        col("Delivery_Delay") > 0,
        "Delayed"
    )
    .otherwise(
        "On Time"
    )
)


delivery_time.show(5)
```

```python
total_delivery = delivery_time.count()


on_time = delivery_time \
    .filter(
        col("Status")=="On Time"
    ) \
    .count()


accuracy = (
    on_time / total_delivery
) * 100


print(
    "Tỷ lệ giao đúng hạn:",
    accuracy,
    "%"
)
```
